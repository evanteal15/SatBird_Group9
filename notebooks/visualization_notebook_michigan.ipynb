```python
import os
import sys
from pathlib import Path

sys.path.append(str(Path().resolve().parent))
sys.path.append(str(Path().resolve().parent.parent))

import numpy as np
import pandas as pd
import json

import matplotlib
matplotlib.use('Agg')

import geopandas as gpd
import matplotlib.pyplot as plt

from shapely.geometry import Point
```

```python
# import seaborn as sns
```

```python
#map of hotspots

path = "../SatBird_data_v3/all_summer_hotspots.csv"

df = pd.read_csv(path)
df = df.drop_duplicates("hotspot_id")

geoDatav = gpd.read_file('https://raw.githubusercontent.com/holtzy/The-Python-Graph-Gallery/master/static/data/US-counties.geojson')

geometry = [Point(xy) for xy in zip(df['lon'], df['lat'])]
gdf = gpd.GeoDataFrame(df, geometry=geometry)   


train = pd.read_csv("../USA_summer_michigan/train_split.csv")
val = pd.read_csv("../USA_summer_michigan/valid_split.csv")
test = pd.read_csv("../USA_summer_michigan/test_split.csv")
gdf["split"] = ""
idx = gdf[gdf["hotspot_id"].isin(list(train["hotspot_id"]))].index
gdf.loc[idx,"split"] = "train"

idx = gdf[gdf["hotspot_id"].isin(list(val["hotspot_id"]))].index
gdf.loc[idx,"split"] = "val"
idx = gdf[gdf["hotspot_id"].isin(list(test["hotspot_id"]))].index
gdf.loc[idx,"split"] = "test"


fig, ax = plt.subplots(figsize =(20,15))
#train_gdf.drop_duplicates(["geometry"]).boundary.plot(ax = ax, alpha = 0.4, edgecolor = "gray")
# geoDatav[~geoDatav["STATE"].isin(["02", "15", "72"])].boundary.plot(ax=ax, alpha = 0.1, edgecolor = "gray" )
geoDatav[geoDatav["STATE"] == "26"].boundary.plot(ax=ax, alpha = 0.4, edgecolor = "gray" )
gdf[gdf["split"]=="train"].plot(ax=ax,marker='o', color='#009E73', markersize=50, label = "train")
gdf[gdf["split"]=="val"].plot(ax=ax, marker='o', color='#56B4E9', markersize=50, label = "val")
gdf[gdf["split"]=="test"].plot(ax=ax, marker='o', color='#E69F00', markersize=50, label = "test")

plt.legend(fontsize=40, markerscale=5, loc='lower left', bbox_to_anchor=(0, 0), frameon=False)
#plt.title("Hotspots")
plt.axis('off')
plt.savefig("hotspots_satbird_summer.png",  bbox_inches='tight')
plt.show()
```
